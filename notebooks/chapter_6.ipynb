Al asignarle el valor que tenia `a` **3**, si cambiamos el valor de `a` a **spam**, `b` sigue referenciando al objeto original, **3**

```python
a = 3
b = a
print("b es ", b)

a = 3
b = a
a = "spam"
print("b es ", b)
```

Los `integers` son objetos inmutables

```python
a = 3
b = a
a = a + 2
print("b es ", b)
```

### OBJETOS MUTABLES

#### Listas 

A diferencia de los `integers` las `listas` son mutables, `M` referencia a `L` y si es modificada, `M` también

```python
L = [2, 3, 4]
M = L
L[0] = 24

print("L = ", L)
print("M = ", M)
```

Usando `[:]` copiamos el objeto, en este caso, si modificamos `L`, `M` no resulta afectada

```python
L = [2, 3, 4]
M = L[:]
L[0] = 24

print("L = ", L)
print("M = ", M)
```

#### Referencia compartida e igualdad

- Operador `==` compara que valgan lo mismo
- Operador `is` compara si referencian al mismo objeto

```python
L = [2, 3, 4]
M = L

print(L == M)
print(L is M)
```

Como referencian a objetos con el mismo valor pero no usan el mismo espacio de memoria, `==` es **True** pero `is` es **False**

```python
L = [1, 2, 3]
M = [1, 2, 3]

print("L == M ->", L == M)
print("L is M ->", L is M)
```

#### CACHE

Los integers si con el mismo valor, usan el mismo espacio de memoria por eso `==` y `is` son **True**

```python
X = 42
Y = 42

print(X == Y)
print(X is Y)
```

`sys` cuenta la cantidad de veces que se nombra lo que está en parentesis

```python
import sys
print(sys.getrefcount([1, 2, 3]))
print(sys.getrefcount(1))
```
