# Diccionarios

Un diccionario es una estructura de datos que almacena pares de elemenentos de forma clave-valor.

La clave tiene que ser un objeto __inmutable__ como un string y el valor puede suer cualquier otro objeto de python.

## Declaración

los diccionarios se declaran con llaves `{}` y los pares se separan con comas

diccionario = {calve1: valor1, clave2: valor2 ...}

- Se accede a los elementos por su clave o key, no por la posición que ocupan.
- Son colecciones no ordenadas de objetos de cualquier tipo, y el tipo de los elementos que lo forman puede ser distinto.
- Poseen longitud variable y pueden anidarse.

```python
diccionario = { "hola": "hi",
                "adios": "bye",
                "casa": "home",
                "perro": "dog"} # clave y valor son strings

print("diccionario ->", diccionario)


telefonos = {   "Juan": 623123324,
                "Pedro": 512623423,
                "Ana": 624852312,
                "Maria": 648238275} # la clave es un string y el valor un integer

print("telefonos ->", telefonos)


meses = {   1: "Enero",
            2: "Febrero",
            3: "Marzo",
            4: "Abril",
            5: "Mayo",
            6: "Junio",
            7: "Julio",
            8: "Agosto",
            9: "Septiembre",
            10: "Octubre",
            11: "Noviembre",
            12: "Diciembre"} # la calve es un integer y el valor un string

print("meses ->", meses)
print("mes nº 8 ->", meses[8])

# declarar diccionarios
codigos = {}
codigos["Programacion"] = "MP153"
codigos["Matematicas"] = "MP532"
codigos["Lengua"] = "MP953"

print("codigos ->", codigos)
```

## Acceso a elementos individuales

Si no encuentra la clave devuelve una excepción `KeyError`

```python
print(codigos["Programacion"])
print(codigos["Historia"])
```

## Recorrido de los elementos

No es una secciencia pero se puede adaptar para poder recorrerla.

Podemos acceder a las claves mediante el método `keys()` y a los valores mediante el método `values()`.

```python
codigos = {}
codigos["Programacion"] = "MP0485"
codigos["Sistemas informaticos"] = "MP0483"
codigos["Contornos de desenvolvemento"] = "MP0487"
codigos["Bases de datos"] = "MP0484"
codigos["Linguaxes de marcas"] = "MP0373"
codigos["Formacion e orientacion laboral"] = "MP0617"

for clave in codigos: # se llama al método __iter__() del diccionario de forma implícita
    print(clave, codigos[clave])
```

```python
for clave in codigos.keys():
    print(clave, codigos[clave])
```

```python
for valor in codigos.values():
    print(valor)
```

#### `items`()

Con `items()` devuelve todas las tuplas (key, values).

```python
for clave, valor in codigos.items(): # Devuelve una tupla y desempaqueta
    print(clave, valor)
```

```python
for tupla in codigos.items(): # Devuelve una tupla
    print(tupla)  
```

#### `get()`

Devuelve la clave del elemento que pidas, si no está en el diccionario, devuelve `None`

```python
D.get('spam')
# 2
```

```python
print(D.get('toast'))       # Una clave que no existe
# None
```

#### update()

El metodo `update` une las claves y valores de un diccionaroi en otro, sobreescribiendo valores de la misma clave si hay conflicto.

```python
# D = {'eggs': 3, 'spam': 2, 'ham': 1}
D2 = {'toast':4, 'muffin':5, 'ham': 10}
D.update(D2)
D
# {'eggs': 3, 'muffin': 5, 'toast': 4, 'spam': 2, 'ham': 10}
```

#### `pop()`

`pop` elimina una clave de un diccionario y devuelve el valor asociado a la clave.

```python
D.pop('muffin')
# 5
D.pop('toast')
# 4
D
# {'eggs': 3, 'spam': 2, 'ham': 10}
```

## Modificación de los elementos



```python
precios_frutas = {
                    "manzana": 2.35,
                    "pera": 3.05,
                    "naranja": 1.75,
                    "limon": 1.95,
                    "melon": 4.85,
                    "sandia": 3.45,
                    "platano": 2.95,
                    "cereza": 5.90,
                    "uva": 4.15,
                    "fresa": 4.25,
}

print (precios_frutas)

precios_frutas["fresa"] = 6.54

print("precio fresa -> ", precios_frutas["fresa"])

for fruta in precios_frutas:
    precios_frutas[fruta] = round(precios_frutas[fruta]) # round redondea los float

print(precios_frutas)
```

### Cambiar diccionarios *in place*

```python
# D = {key: value}
D = {'spam': 2, 'ham': 1, 'eggs': 3}        # construir un diccionario
D

D['ham'] = ['grill', 'bake', 'fry']         # Change entry (value=list)
D
# {'eggs': 3, 'spam': 2, 'ham': ['grill', 'bake', 'fry']}
```

```python
D['spam']    
```

```python
'ham' in D      # Key membership test alternative
# True
```

```python
list(D.keys())      # Crear una lista con las claves del diccionario
# ['eggs', 'spam', 'ham']
```

## Eliminación de un elemento

para eliminar un elemento podemos usar la palabra reservada `del`

```python
precios_frutas = {
                "manzana": 2.35,
                "pera": 3.05,
                "naranja": 1.75,
                "limon": 1.95,
                "melon": 4.85,
                "sandia": 3.45,
                "platano": 2.95,
                "cereza": 5.90,
                "uva": 4.15,
                "fresa": 4.25,
                }

print(precios_frutas)

del precios_frutas["manzana"]

print(precios_frutas)

del precios_frutas

print(precios_frutas)
```
