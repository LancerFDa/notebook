# Diccionarios

Un diccionario es una estructura de datos que almacena pares de elemenentos de forma clave-valor.

La clave tiene que ser un objeto __inmutable__ como un string y el valor puede suer cualquier otro objeto de python.

## Declaración

los diccionarios se declaran con llaves `{}` y los pares se separan con comas

diccionario = {calve1: valor1, clave2: valor2 ...}

```python
diccionario = { "hola": "hi",
                "adios": "bye",
                "casa": "home",
                "perro": "dog"} # clave y valor son strings

print("diccionario ->", diccionario)


telefonos = {   "Juan": 623123324,
                "Pedro": 512623423,
                "Ana": 624852312,
                "Maria": 648238275} # la clave es un string y el valor un integer

print("telefonos ->", telefonos)


meses = {   1: "Enero",
            2: "Febrero",
            3: "Marzo",
            4: "Abril",
            5: "Mayo",
            6: "Junio",
            7: "Julio",
            8: "Agosto",
            9: "Septiembre",
            10: "Octubre",
            11: "Noviembre",
            12: "Diciembre"} # la calve es un integer y el valor un string

print("meses ->", meses)
print("mes nº 8 ->", meses[8])

# declarar diccionarios
codigos = {}
codigos["Programacion"] = "MP153"
codigos["Matematicas"] = "MP532"
codigos["Lengua"] = "MP953"

print("codigos ->", codigos)
```

## Acceso a elementos individuales

Si no encuentra la clave devuelve una excepción `KeyError`

```python
print(codigos["Programacion"])
print(codigos["Historia"])
```

## Recorrido de los elementos

No es una secciencia pero se puede adaptar para poder recorrerla.

Podemos acceder a las claves mediante el método `keys()` y a los valores mediante el método `values()`.

```python
codigos = {}
codigos["Programacion"] = "MP0485"
codigos["Sistemas informaticos"] = "MP0483"
codigos["Contornos de desenvolvemento"] = "MP0487"
codigos["Bases de datos"] = "MP0484"
codigos["Linguaxes de marcas"] = "MP0373"
codigos["Formacion e orientacion laboral"] = "MP0617"

for clave in codigos: # se llama al método __iter__() del diccionario de forma implícita
    print(clave, codigos[clave])
```

```python
for clave in codigos.keys():
    print(clave, codigos[clave])
```

```python
for valor in codigos.values():
    print(valor)
```

```python
for clave, valor in codigos.items(): # Devuelve una tupla y desempaqueta
    print(clave, valor)
```

```python
for tupla in codigos.items(): # Devuelve una tupla
    print(tupla)  
```
